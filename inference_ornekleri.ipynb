# Inference Örnekleri

```python
from transformers import pipeline

# Metin üretimi için pipeline oluşturma
# Donanımına göre device=0 (GPU) veya device=-1 (CPU) parametresi eklenebilir
generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct"
)

# Modele gönderilecek girdi
prompt = "<|system|>\nSen yardımsever bir asistanın.</s>\n<|user|>\nBüyük Dil Modelleri (LLM) nasıl çalışır? Kısaca açıkla.</s>\n<|assistant|>\n"

# Inference işlemi
sonuc = generator(prompt, max_new_tokens=100, return_full_text=False)

print(sonuc[0]["generated_text"])
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_id = "Qwen/Qwen2.5-0.5B-Instruct"

# 1. Tokenizer ve Modeli yükleme
tokenizer = AutoTokenizer.from_pretrained(model_id)
# Bellek optimizasyonu için float16 kullanıyoruz
model = AutoModelForCausalLM.from_pretrained(model_id, torch_dtype=torch.float16)

# 2. Girdiyi Hazırlama (Chat şablonu kullanımı)
soru = "Transformer mimarisinin en önemli bileşeni nedir?"
mesajlar = [
    {"role": "system", "content": "Sen bir yapay zeka araştırmacısısın. Kısa ve net cevaplar verirsin."},
    {"role": "user", "content": soru},
]

# Sohbet geçmişini modelin anladığı özel tokenlarla formata sokma
prompt = tokenizer.apply_chat_template(mesajlar, tokenize=False, add_generation_prompt=True)

# Metni sayılara (tensor) çevirme
girdiler = tokenizer(prompt, return_tensors="pt")

# 3. Çıkarım (Inference)
with torch.no_grad(): # Sadece çıkarım yapacağımız için gradient hesaplamayı kapatıyoruz
    ciktilar = model.generate(
        **girdiler,
        max_new_tokens=150,
        temperature=0.7, # Yaratıcılık/Rastgelelik seviyesi
        do_sample=True
    )

# 4. Çıktıyı insan diline geri çevirme (Decoding)
yanit = tokenizer.decode(ciktilar[0], skip_special_tokens=True)

print(yanit)
```
