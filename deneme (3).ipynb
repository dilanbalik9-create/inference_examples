# Text Generation

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-0.5B-Instruct")
```

```python
import torch
from transformers import AutoModelForCausalLM

model = AutoModelForCausalLM.from_pretrained("Qwen/Qwen2.5-0.5B-Instruct", dtype=torch.float16).to("cuda")

text = "This is my second internship but still I"

inputs = tokenizer(text, return_tensors="pt").to("cuda")

with torch.no_grad(): # Sadece çıkarım yapacağımız için gradient hesaplamayı kapatıyoruz. Gradyanlar sadece yapay zeka eğitilirken lazımdır
    ciktilar = model.generate(
        **inputs,
        max_new_tokens=150,
        temperature=0.7, # Yaratıcılık/Rastgelelik seviyesi
        do_sample=True # ihtimaller havuzundan olasılığa göre seçim yapmasını sağlar. temperature için gereklidir.
)
```

```python
sonuc = tokenizer.decode(ciktilar[0], skip_special_tokens=True)
print(sonuc)
```
