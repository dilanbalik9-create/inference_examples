```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-0.5B-Instruct")
inputs = tokenizer("It's my second internship but still ", return_tensors="pt").to("cuda")
```

```python
import torch
from transformers import AutoModelForCausalLM

model = AutoModelForCausalLM.from_pretrained("Qwen/Qwen2.5-0.5B-Instruct", dtype=torch.float16).to("cuda")
```

```python
outputs = model.generate(**inputs, max_new_tokens=20, do_sample = True, top_p =0.95, top_k = 20, temperature = 0.7)
tokenizer.batch_decode(outputs, skip_special_tokens=True)
```
